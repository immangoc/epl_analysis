```python
season = matches_all[matches_all['season'] == 2526].copy()
season['date'] = pd.to_datetime(season['date'])
season = season.sort_values('date').reset_index(drop=True)
```

```python
season['date'].dtype
```

```python
def normalize_pos(pos):
    return ','.join(sorted(pos.split(',')))

players_epl['Pos'] = players_epl['Pos'].apply(normalize_pos)
players_epl['Pos'].unique()
```

```python
season['home_result'] = season['home_points'].map({3: 'W', 1: 'D', 0: 'L'})
season['away_result'] = season['away_points'].map({3: 'W', 1: 'D', 0: 'L'})
season[['home_team','away_team','home_goals','away_goals','home_result','away_result']].head()
```

```python
home_view = season[['date','home_team','away_team','home_points','home_goals','away_goals','home_xg','away_xg']].copy()
home_view.columns = ['date','team','opponent','points','goals_for','goals_against','xg_for','xg_against']
home_view['venue'] = 'Home'

home_view.head()
```

```python
away_view = season[['date', 'away_team', 'home_team', 'away_points', 'away_goals', 'home_goals', 'away_xg', 'home_xg']].copy()
away_view.columns = ['date', 'team', 'opponent', 'points', 'goals_for', 'goals_against', 'xg_for', 'xg_against']
away_view['venue'] = 'Away'

away_view.head()
```

```python
team_matches = pd.concat([home_view, away_view], ignore_index=True)
team_matches = team_matches.sort_values(['team', 'date']).reset_index(drop=True)
team_matches.shape
```

```python
len(team_matches) == len(home_view) + len(away_view)
```

```python
team_matches['team'].value_counts()
```

```python
team_matches['result'] = team_matches['points'].map({3: 'W', 1: 'D', 0: 'L'})
team_matches['goal_diff'] = team_matches['goals_for'] - team_matches['goals_against']
team_matches['xg_diff'] = team_matches['xg_for'] - team_matches['xg_against']
```

```python
team_matches['matchweek'] = team_matches.groupby('team').cumcount() + 1
team_matches['cum_points'] = team_matches.groupby('team')['points'].cumsum()
```

```python
team_matches[team_matches['team'] == 'Arsenal'][['date','opponent','venue','result','points','matchweek','cum_points']].head(10)
```

```python
team_matches['form_pts_last5'] = (
    team_matches.groupby('team')['points']
    .rolling(5, min_periods=1)
    .sum()
    .reset_index(level=0, drop=True)
)
```

```python
team_matches[team_matches['team'] == 'Arsenal'][['date','opponent','result','points','form_pts_last5']].head(10)
```

```python
players_epl['Ast_per90'] = players_epl['Ast'] / players_epl['Min'] * 90
players_epl['Sh_per90'] = players_epl['Sh'] / players_epl['Min'] * 90
players_epl['Gls_per90'] = players_epl['Gls'] / players_epl['Min'] * 90
```

```python
players_epl[players_epl['Min'] >= 450][['Player','Squad','Min','Gls','Gls_per90','Ast','Ast_per90']].sort_values('Gls_per90', ascending=False).head(10)
```

```python
[c for c in players_epl.columns if 'per90' in c]
```

```python
standings = team_matches.groupby('team').agg(
    MP=('points','count'),
    W=('result', lambda x: (x=='W').sum()),
    D=('result', lambda x: (x=='D').sum()),
    L=('result', lambda x: (x=='L').sum()),
    GF=('goals_for','sum'),
    GA=('goals_against','sum'),
    Pts=('points','sum'),
).reset_index()

standings['GD'] = standings['GF'] - standings['GA']
standings = standings.sort_values(['Pts','GD','GF'], ascending=False).reset_index(drop=True)
standings.insert(0, 'Rank', standings.index + 1)
standings
```

```python
players_epl.to_csv('players_clean.csv', index=False)
team_matches.to_csv('team_matches_clean.csv', index=False)
standings.to_csv('standings.csv', index=False)
```
